```python
import pandas as pd

df = pd.read_csv("orginal.csv")
teryt = pd.read_csv("TERC_Adresowy_2026-04-16.csv", sep=";")

# =========================
# 1. DF - KODY Z DZIAŁKI
# =========================
df["kod_gminy"] = df["dzi_id_dzialki"].str.split("_").str[0].str[:6]  # np. 142706
df["kod_powiat"] = df["kod_gminy"].str[:4]  # np. 1427

# =========================
# 2. TERYT - NORMALIZACJA
# =========================
teryt["WOJ"] = teryt["WOJ"].fillna(0).astype(int).astype(str).str.zfill(2)
teryt["POW"] = teryt["POW"].fillna(0).astype(int).astype(str).str.zfill(2)
teryt["GMI"] = teryt["GMI"].fillna(0).astype(int).astype(str).str.zfill(2)

# pełne kody
teryt["kod_gminy"] = teryt["WOJ"] + teryt["POW"] + teryt["GMI"]
teryt["kod_powiat"] = teryt["WOJ"] + teryt["POW"]

# =========================
# 3. POWIATY
# =========================
powiaty = teryt[teryt["GMI"] == "00"].copy()

# =========================
# 4. GMINY
# =========================
gminy = teryt[teryt["GMI"] != "00"].copy()

# =========================
# 5. MERGE - GMINA
# =========================
df = df.merge(
    gminy[["kod_gminy", "NAZWA"]],
    on="kod_gminy",
    how="left"
)

df.rename(columns={"NAZWA": "gmina"}, inplace=True)

# =========================
# 6. MERGE - POWIAT
# =========================
df = df.merge(
    powiaty[["kod_powiat", "NAZWA"]],
    on="kod_powiat",
    how="left"
)

df.rename(columns={"NAZWA": "powiat"}, inplace=True)

# =========================
# 7. ZAPIS
# =========================
df.to_csv("zzgmina.csv", index=False)

#print(df[["gid", "kod_gminy", "gmina", "powiat", 'dzi_id_dzialki']])
print(df.head())
```
